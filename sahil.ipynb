```python
# Define class : vehicle
class vehicle:
  company_name = "TATA" # class variable / class attribute
  
# define constructor / Special method / Instance method
  def __init__(self,name,age):
    self.name = name # Instance varible / Instance attribute
    self.age = age  # Instance varible / Instance attribute

  # Define Method
  def sound(self):
    print(f'{self.name} and {self.age}')
    return "beeeeep"
```

```python
# define object
tata = vehicle('sahil',12)
```

```python
# fecthing the sound() method using .operator
tata.sound()
```
